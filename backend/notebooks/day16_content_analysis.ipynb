```python
# Cell 1: Setup

import sys
sys.path.append('..')

from bufferiq.ml.content.intelligence import ContentIntelligenceService
from bufferiq.ml.content.preprocessing import TextCleaner
from bufferiq.ml.content.sentiment import SentimentAnalyzer
from bufferiq.ml.content.readability import ReadabilityAnalyzer
from bufferiq.ml.content.optimizer import ContentOptimizer

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')

%matplotlib inline
```

```python
# Cell 2: Text Preprocessing

cleaner = TextCleaner()

text = "Excited to share our AI project! 🚀 Check https://example.com #AI #Innovation @team"

result = cleaner.clean(text)

print(f"Original: {result.original}")
print(f"Cleaned: {result.cleaned}")
print(f"Hashtags: {result.hashtags}")
print(f"Mentions: {result.mentions}")
print(f"URLs: {result.urls}")
print(f"Emojis: {result.emojis}")
```

```python
# Cell 3: Sentiment Analysis

analyzer = SentimentAnalyzer()

texts = [
    "I love this product!",
    "This is terrible.",
    "It's okay, nothing special.",
    "Absolutely amazing experience!"
]

results = []

for text in texts:
    result = analyzer.analyze(text)

    results.append({
        'text': text,
        'sentiment': result.sentiment.value,
        'confidence': result.confidence,
        'polarity': result.polarity
    })

df = pd.DataFrame(results)

print(df)
```

```python
# Cell 4: Readability Analysis

readability_analyzer = ReadabilityAnalyzer()

texts = [
    "The cat sat on the mat.",
    "Implementation of sophisticated algorithms necessitates comprehensive understanding."
]

for text in texts:
    scores = readability_analyzer.analyze(text)

    print(f"\nText: {text}")
    print(f"Flesch Reading Ease: {scores.flesch_reading_ease:.1f}")
    print(f"Grade Level: {scores.average_grade_level:.1f}")
    print(f"Difficulty: {scores.reading_difficulty}")
```

```python
# Cell 5: Content Optimization

optimizer = ContentOptimizer()

text = "Quick update"

result = optimizer.optimize(text, "linkedin")

print(f"Overall Score: {result.overall_score:.1f}/100")
print(f"Predicted Lift: {result.predicted_engagement_lift:.1f}%")
print(f"Best Platform: {result.best_platform}")

print("\nSuggestions:")

for sug in result.suggestions:
    print(f" [{sug.priority}] {sug.type}: {sug.impact}")
```

```python
# Cell 6: Full Intelligence Analysis

service = ContentIntelligenceService()

text = "Excited to share our new AI project! 🚀 #AI #Innovation"

analysis = service.analyze_content(text, "linkedin")

print("Sentiment:", analysis['sentiment']['sentiment'])
print("Quality Score:", analysis['quality']['score'])
print("Optimization Score:", analysis['optimization']['overall_score'])

print("\nFeatures:")
print(f" Words: {analysis['features']['word_count']}")
print(f" Hashtags: {analysis['features']['hashtag_count']}")
print(f" Emojis: {analysis['features']['emoji_count']}")
```

```python
# Cell 7: Batch Analysis

posts = [
    {"text": "Great post about AI!"},
    {"text": "Data science is amazing"},
    {"text": "Machine learning update"},
]

results = service.analyze_batch(posts, "linkedin")

summary = []

for r in results:
    summary.append({
        'text': r['text'][:30] + '...',
        'sentiment': r['sentiment']['sentiment'],
        'quality': r['quality']['score'],
        'opt_score': r['optimization']['overall_score']
    })

df = pd.DataFrame(summary)

print(df)
```

```python
# Cell 8: Visualization

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

df['quality'].plot(kind='bar', ax=axes[0], color='skyblue')
axes[0].set_title('Quality Scores')
axes[0].set_ylabel('Score')
axes[0].set_ylim(0, 100)

df['opt_score'].plot(kind='bar', ax=axes[1], color='lightcoral')
axes[1].set_title('Optimization Scores')
axes[1].set_ylabel('Score')
axes[1].set_ylim(0, 100)

plt.tight_layout()
plt.show()
```
