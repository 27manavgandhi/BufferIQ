```python
# Cell 1: Setup

import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from bufferiq.ml.features.pipeline import FeatureEngineeringPipeline
from bufferiq.ml.features.temporal import TemporalFeatureExtractor
from bufferiq.ml.features.content import ContentFeatureExtractor
from bufferiq.ml.features.nlp import NLPFeatureExtractor

%matplotlib inline
sns.set_style('whitegrid')
```

```python
# Cell 2: Load Sample Data

df = pd.DataFrame({
    'published_at': [
        '2024-01-01T10:00:00Z',
        '2024-01-02T14:00:00Z',
        '2024-01-03T18:00:00Z',
    ],
    'content': [
        'Great post about #AI and #ML! Check out https://example.com',
        'Simple update',
        'Question: What do you think about this? @user',
    ],
    'platform': ['linkedin', 'twitter', 'bluesky']
})

df.head()
```

```python
# Cell 3: Extract Temporal Features

temporal = TemporalFeatureExtractor()
temporal_features = temporal.extract(df)

print(f"Extracted {len(temporal_features.columns)} temporal features")
temporal_features.head()
```

```python
# Cell 4: Extract Content Features

content = ContentFeatureExtractor()
content_features = content.extract(df)

print(f"Extracted {len(content_features.columns)} content features")
content_features.head()
```

```python
# Cell 5: Extract NLP Features

nlp = NLPFeatureExtractor()
nlp_features = nlp.extract(df)

print(f"Extracted {len(nlp_features.columns)} NLP features")
nlp_features.head()
```

```python
# Cell 6: Full Pipeline Extraction

pipeline = FeatureEngineeringPipeline()
all_features = await pipeline.extract_features(df)

print(f"Total features extracted: {len(all_features.columns)}")
all_features.head()
```

```python
# Cell 7: Feature Statistics

stats = pipeline.get_feature_stats()
for key, value in stats.items():
    print(f"{key}: {value}")
```

```python
# Cell 8: Visualize Feature Distributions

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

temporal_features['hour'].hist(ax=axes[0, 0])
axes[0, 0].set_title('Hour of Day Distribution')

temporal_features['day_of_week'].hist(ax=axes[0, 1])
axes[0, 1].set_title('Day of Week Distribution')

temporal_features['is_weekend'].value_counts().plot(kind='bar', ax=axes[1, 0])
axes[1, 0].set_title('Weekend Posts')

temporal_features['is_business_hours'].value_counts().plot(kind='bar', ax=axes[1, 1])
axes[1, 1].set_title('Business Hours Posts')

plt.tight_layout()
plt.show()
```
