```python
# Cell 1: Setup

import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from bufferiq.ml.training.data_preparation import DataPreparation
from bufferiq.ml.training.experiment_tracker import ExperimentTracker
from bufferiq.ml.training.model_registry import ModelRegistry
from bufferiq.ml.training.checkpoint import Checkpoint

%matplotlib inline
sns.set_style('whitegrid')
```

```python
# Cell 2: Load Sample Data

df = pd.DataFrame({
    'feature1': list(range(200)),
    'feature2': list(range(200, 400)),
    'feature3': list(range(400, 600)),
    'engagement_rate': [10 + i*0.1 for i in range(200)],
    'published_at': pd.date_range('2024-01-01', periods=200),
    'platform': ['linkedin'] * 100 + ['twitter'] * 100
})

df.head()
```

```python
# Cell 3: Data Preparation

prep = DataPreparation(
    test_size=0.2,
    validation_size=0.1,
    time_based_split=True
)

X_train, X_val, X_test, y_train, y_val, y_test = prep.split_data(
    df,
    target_column='engagement_rate',
    feature_columns=['feature1', 'feature2', 'feature3'],
    time_column='published_at'
)

print(f"Train size: {len(X_train)}")
print(f"Val size: {len(X_val)}")
print(f"Test size: {len(X_test)}")
```

```python
# Cell 4: Data Quality Validation

quality = prep.validate_data_quality(df)

print(f"Total samples: {quality['total_samples']}")
print(f"Missing values: {quality['missing_values']}")
print(f"Warnings: {quality['warnings']}")
```

```python
# Cell 5: Experiment Tracking

tracker = ExperimentTracker(experiment_name='notebook_experiment')

tracker.log_params({
    'test_size': 0.2,
    'validation_size': 0.1,
    'model_type': 'random_forest'
})

tracker.log_metrics({
    'train_size': len(X_train),
    'val_size': len(X_val),
    'test_size': len(X_test)
})

print("Experiment logged!")
```

```python
# Cell 6: Visualize Data Distribution

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

y_train.hist(ax=axes[0], bins=20)
axes[0].set_title('Training Target Distribution')
axes[0].set_xlabel('Engagement Rate')

df['platform'].value_counts().plot(kind='bar', ax=axes[1])
axes[1].set_title('Platform Distribution')

plt.tight_layout()
plt.show()
```
