```python
# Cell 1: Setup

import numpy as np
import matplotlib.pyplot as plt

from bufferiq.ml.experiments import (
    ExperimentDesigner,
    StatisticalAnalyzer
)

from bufferiq.ml.experiments.design.designer import (
    Variant,
    MetricType
)

print("✓ Imports successful")
```

```python
# Cell 2: Design Experiment

designer = ExperimentDesigner()

variants = [
    Variant(
        "control",
        "Control",
        "Original",
        0.5,
        {},
        True
    ),
    Variant(
        "treatment",
        "Treatment",
        "New",
        0.5,
        {}
    )
]

config = designer.design(
    name="Headline Test",
    description="Test AI headlines",
    variants=variants,
    platform="linkedin",
    primary_metric=MetricType.ENGAGEMENT_RATE,
    baseline_rate=0.05,
    mde=0.10
)

print(f"Experiment: {config.name}")
print(f"Required sample size: {config.required_sample_size:,} per variant")
print(f"Estimated duration: {config.estimated_duration_days} days")
```

```python
# Cell 3: Simulate Data

np.random.seed(42)

n = 1000

control_data = np.random.binomial(
    1,
    0.05,
    n
)

treatment_data = np.random.binomial(
    1,
    0.055,
    n
)

print(f"Control mean: {control_data.mean():.3f}")
print(f"Treatment mean: {treatment_data.mean():.3f}")

print(
    f"Relative lift: "
    f"{(treatment_data.mean() - control_data.mean()) / control_data.mean():.1%}"
)
```

```python
# Cell 4: Analyze Results

analyzer = StatisticalAnalyzer()

result = analyzer.analyze(
    control_data=control_data,
    treatment_data=treatment_data,
    metric_type=MetricType.ENGAGEMENT_RATE
)

print(f"Test type: {result.test_type}")
print(f"P-value: {result.p_value:.4f}")
print(f"Is significant: {result.is_significant}")
print(f"Effect size: {result.effect_size:.3f}")

print(
    f"95% CI: "
    f"[{result.ci_lower:.4f}, {result.ci_upper:.4f}]"
)
```

```python
# Cell 5: Visualization

labels = ['Control', 'Treatment']
means = [
    control_data.mean(),
    treatment_data.mean()
]

plt.figure(figsize=(8, 5))

plt.bar(labels, means)

plt.ylabel('Engagement Rate')
plt.title('Experiment Variant Comparison')

plt.grid(axis='y', alpha=0.3)

plt.show()
```
